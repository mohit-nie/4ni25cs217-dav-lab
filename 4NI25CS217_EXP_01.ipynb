```python
import numpy as np
import pandas as pd
```

```python
arr = np.array([10, 20, 30, 40, 50])
print('Array: ', arr)
```

```python
print('Mean: ', arr.mean())
```

```python
print("Reshaped 2d:\n",arr.reshape(5,1))
```

```python
print("Sliced (index 1 to 3):", arr[1:4])
```

```python
data={'Name':['yayti','yojitha','khan','mohammed'],
      'Marks':[88,72,91,65],
      'Branch':['CSE','ECE','CSE','ISE']}
df=pd.DataFrame(data)
print(df)
```

```python
print('\nMarks column:\n',df)
```

```python
print('\nRows where Marks > 70:\n', df[df['Marks'] > 70])
```
